```python
import pandas as pd

df = pd.read_csv("data/dataset-edad-ingresos.csv")
df
```

```python
# intuitive -> the first 1000

import seaborn as sns
first_one_thousand = df.iloc[0:1000,:]
ax = sns.scatterplot(data=first_one_thousand, y='Ingresos (USDx1000)', x="Edad")
```

```python
# intuitive -> random 1000

import seaborn as sns
random_thousand = df.sample(1000)
ax = sns.scatterplot(data=random_thousand, y='Ingresos (USDx1000)', x="Edad")
```

```python
import numpy as np

for aux in [df, first_one_thousand, random_thousand]:
    print(f'mean: {aux["Edad"].mean():.1f} | std: {aux["Edad"].std():.1f}')
    print(f'mean: {aux["Ingresos (USDx1000)"].mean():.1f} | std: {aux["Ingresos (USDx1000)"].std():.1f}')
```
