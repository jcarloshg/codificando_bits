# Graph Density

- El gráfico de violín muestra la **densidad** de los datos: el ancho de la figura en cada valor indica cuántos datos se concentran ahí.
- Cada variable se grafica por separado en un grid de 3 filas x 1 columna para poder comparar su forma.
- La caja interior (`inner="box"`) marca la mediana y el rango intercuartil (IQR).


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/data-histogramas.csv")
df.describe()
```

```python
variables = ["var_1", "var_2", "var_3"]

fig, axes = plt.subplots(3, 1, figsize=(6, 12))

for ax, var in zip(axes, variables):
    sns.violinplot(data=df[var], ax=ax, color="skyblue", inner="box")
    ax.set_xlabel("Variable")
    ax.set_ylabel("Valores")
    ax.set_xticks([0], labels=[var])
    ax.set_title(var)

plt.tight_layout()
plt.show()
```
